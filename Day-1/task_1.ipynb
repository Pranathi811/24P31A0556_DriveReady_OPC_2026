```python
from langchain_groq import ChatGroq
```

```python
import os
from google.colab import userdata

# Retrieve the Groq API key from Colab Secrets (add it to the Secrets tab on the left sidebar)
try:
    os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
except Exception:
    print("Please add GROQ_API_KEY to your Colab Secrets.")

# Initialize the LLM with a valid Groq model
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
SYSTEM = """
You are an AI Trading Research Assistant.

Your task is to analyze financial markets and provide educational
trading analysis, not guaranteed predictions.

Analyze the market using the following framework:

1. Market Trend: Identify bullish, bearish, or sideways trend.
2. Technical Analysis: Analyze RSI, MACD, moving averages, and support/resistance.
3. Entry Price: Suggest a hypothetical entry zone if data supports it.
4. Stop Loss: Identify a potential invalidation level.
5. Target: Suggest potential target levels.
6. Risk-Reward Ratio: Calculate the hypothetical risk/reward ratio.
7. Trade Decision: BUY, SELL, or WAIT based on available data.
8. Reasoning: Explain the decision clearly.
9. Risk Management: Risk no more than 1% of capital per hypothetical trade.

Rules:
- Never invent live market prices.
- If market data is missing, ask for it.
- Always consider risk before profit.
- Do not guarantee returns.
- Treat this as paper trading and educational analysis.
- Give answers in a structured, concise format.
"""
```

```python
def ask(question):
    return llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ]).content
```

```python
question = """
Analyze NIFTY 50 for a hypothetical intraday trade.

Assume:
- Capital: ₹1,00,000
- Maximum risk: 1% per trade
- Trading timeframe: 5-minute chart

Provide:
- Market trend
- Entry zone
- Stop loss
- Target price
- Position sizing
- Risk-reward ratio
- Final decision: BUY, SELL, or WAIT

If real market data is unavailable, do not invent prices.
"""

print(ask(question))
```
